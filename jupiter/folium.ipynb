```python
# Step 1: Create the Base Map
import folium

# Define center coordinates for San Francisco
latitude = 37.7749
longitude = -122.4194

# Create a base map
sf_map = folium.Map(location=[latitude, longitude],
                    zoom_start=8,
                    tiles="Esri WorldImagery") # Esri WorldImagery

# Display the map
sf_map
```
